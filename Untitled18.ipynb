```python
# Load the new dataset, convert dates, set index, sort values, and remove missing data.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
from arch import arch_model

df = pd.read_csv("VaR Portfolio.csv")

df["Date"] = pd.to_datetime(df["Date"])
df = df.set_index("Date")
df = df.sort_index()
df = df.dropna()

df.head()
```

```python
# Calculate daily log returns and display summary statistics.

log_returns = np.log(df / df.shift(1))
log_returns = log_returns.dropna()

log_returns.describe()
```

```python
# Create an equally weighted portfolio.

weights = np.repeat(1 / log_returns.shape[1], log_returns.shape[1])
portfolio_returns = log_returns.dot(weights)

portfolio_returns.head()
```

```python
# Display descriptive statistics of the portfolio returns.

portfolio_returns.describe()
```

```python
# Plot the distribution of daily portfolio log returns.

plt.figure(figsize=(10,6))
plt.hist(portfolio_returns, bins=50)

plt.title("Distribution of Daily Portfolio Log Returns")
plt.xlabel("Daily Log Return")
plt.ylabel("Frequency")

plt.show()
```

```python
# Calculate the correlation matrix between asset log returns.

correlation_matrix = log_returns.corr()

correlation_matrix
```

```python
# Visualize the correlation matrix as a heatmap.

plt.figure(figsize=(10,8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix of Daily Log Returns")
plt.show()
```

```python
# Define rolling window and normal quantiles for VaR calculations.

window = 252   # 1 trading year

confidence95 = 0.95
confidence99 = 0.99

z95 = abs(norm.ppf(1 - confidence95))
z99 = abs(norm.ppf(1 - confidence99))
```

```python
# Calculate rolling Delta-Normal VaR using zero mean because daily mean returns are noisy.

rolling_sigma = portfolio_returns.rolling(window).std()

rolling_delta_var95 = z95 * rolling_sigma
rolling_delta_var99 = z99 * rolling_sigma

rolling_delta_var95.tail()
```

```python
# Calculate rolling Delta-Normal VaR using zero mean because daily mean returns are noisy.

rolling_sigma = portfolio_returns.rolling(window).std()

rolling_delta_var95 = z95 * rolling_sigma
rolling_delta_var99 = z99 * rolling_sigma

rolling_delta_var95.tail()
```

```python
# Plot portfolio returns with rolling Delta-Normal VaR estimates.

plt.figure(figsize=(12,6))

plt.plot(portfolio_returns, label="Portfolio Returns")
plt.plot(-rolling_delta_var95, linestyle="--", label="Delta-Normal VaR 95%")
plt.plot(-rolling_delta_var99, linestyle="--", label="Delta-Normal VaR 99%")

plt.title("Portfolio Returns with Rolling Delta-Normal VaR")
plt.xlabel("Date")
plt.ylabel("Daily Log Return")
plt.legend()
plt.show()
```

```python
# Calculate rolling Historical VaR using empirical quantiles over a 252-day window.

rolling_historical_var95 = -portfolio_returns.rolling(window).quantile(0.05)
rolling_historical_var99 = -portfolio_returns.rolling(window).quantile(0.01)

rolling_historical_var95.tail()
```

```python
# Plot portfolio returns with rolling Historical VaR estimates.

plt.figure(figsize=(12,6))

plt.plot(portfolio_returns, label="Portfolio Returns")
plt.plot(-rolling_historical_var95, linestyle="--", label="Historical VaR 95%")
plt.plot(-rolling_historical_var99, linestyle="--", label="Historical VaR 99%")

plt.title("Portfolio Returns with Rolling Historical VaR")
plt.xlabel("Date")
plt.ylabel("Daily Log Return")
plt.legend()
plt.show()
```

```python
# Convert portfolio returns into percentages for the GARCH model.

garch_returns = portfolio_returns * 100
```

```python
# Estimate rolling GARCH(1,1) VaR using a normal distribution and zero mean.

rolling_garch_var95 = []
rolling_garch_var99 = []

for i in range(window, len(garch_returns)):

    sample = garch_returns.iloc[i-window:i]

    model = arch_model(
        sample,
        vol="Garch",
        p=1,
        q=1,
        mean="Zero",
        dist="normal"
    )

    result = model.fit(disp="off")
    forecast = result.forecast(horizon=1)

    sigma_forecast = np.sqrt(forecast.variance.values[-1, 0])

    var95 = (z95 * sigma_forecast) / 100
    var99 = (z99 * sigma_forecast) / 100

    rolling_garch_var95.append(var95)
    rolling_garch_var99.append(var99)
```

```python
# Convert rolling GARCH VaR lists into time series with dates.

rolling_garch_var95 = pd.Series(
    rolling_garch_var95,
    index=portfolio_returns.index[window:]
)

rolling_garch_var99 = pd.Series(
    rolling_garch_var99,
    index=portfolio_returns.index[window:]
)

rolling_garch_var95.tail()
```

```python
# Plot portfolio returns with rolling GARCH VaR estimates.

plt.figure(figsize=(12,6))

plt.plot(portfolio_returns, label="Portfolio Returns")
plt.plot(-rolling_garch_var95, linestyle="--", label="GARCH VaR 95%")
plt.plot(-rolling_garch_var99, linestyle="--", label="GARCH VaR 99%")

plt.title("Portfolio Returns with Rolling GARCH VaR")
plt.xlabel("Date")
plt.ylabel("Daily Log Return")
plt.legend()
plt.show()
```

```python
# Compare the three rolling VaR methods at the 95% confidence level.
# This version forces the x-axis to start from the beginning of the dataset.

plt.figure(figsize=(12,6))

plt.plot(rolling_delta_var95, label="Delta-Normal VaR 95%")
plt.plot(rolling_historical_var95, label="Historical VaR 95%")
plt.plot(rolling_garch_var95, label="GARCH VaR 95%")

plt.title("Comparison of Rolling 1-Year VaR 95%")
plt.xlabel("Date")
plt.ylabel("VaR")

# Force the chart to show the full data period from 2018
plt.xlim(portfolio_returns.index.min(), portfolio_returns.index.max())

plt.legend()
plt.show()
```

```python
# Compare the three rolling VaR methods at the 99% confidence level.
# This version forces the x-axis to start from the beginning of the dataset.

plt.figure(figsize=(12,6))

plt.plot(rolling_delta_var99, label="Delta-Normal VaR 99%")
plt.plot(rolling_historical_var99, label="Historical VaR 99%")
plt.plot(rolling_garch_var99, label="GARCH VaR 99%")

plt.title("Comparison of Rolling 1-Year VaR 99%")
plt.xlabel("Date")
plt.ylabel("VaR")

# Force the chart to show the full data period from 2018
plt.xlim(portfolio_returns.index.min(), portfolio_returns.index.max())

plt.legend()
plt.show()
```

```python
# Combine all VaR estimates into one comparison table.

comparison_results = pd.DataFrame({
    "Delta-Normal VaR 95%": rolling_delta_var95,
    "Historical VaR 95%": rolling_historical_var95,
    "GARCH VaR 95%": rolling_garch_var95,
    "Delta-Normal VaR 99%": rolling_delta_var99,
    "Historical VaR 99%": rolling_historical_var99,
    "GARCH VaR 99%": rolling_garch_var99
})

comparison_results = comparison_results.dropna()

comparison_results.tail()
```

```python
# Display summary statistics for all VaR methods.

comparison_results.describe()
```

```python
# Create a final summary table with mean and maximum VaR for each method.

final_summary = pd.DataFrame({
    "Mean VaR": comparison_results.mean(),
    "Maximum VaR": comparison_results.max()
})

final_summary
```
